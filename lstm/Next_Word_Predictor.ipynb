```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts([docs])
```

```python
word_index=tokenizer.word_index
```

```python
word_counts=tokenizer.word_counts
```

```python
input_sequences=[]
y=[]
for sentence in docs.split("\n"):
  tokenized_sentence=tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1 , len(tokenized_sentence)):
    print(f'{tokenized_sentence[:i]}---> {tokenized_sentence[i]}')
    input_sequences.append(tokenized_sentence[:i])
    y.append(tokenized_sentence[i])
```

```python
max_len=max([len(i) for i in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences=pad_sequences(input_sequences , maxlen=max_len , padding='pre')
```

```python
x=padded_input_sequences
```

```python
x.shape
```

```python
len(word_index)
```

```python
from tensorflow.keras.utils import to_categorical
y=to_categorical(y,num_classes=439)
```

```python
y.shape
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Embedding , LSTM , Dense
```

```python
model=Sequential()

model.add(Embedding(439 , 120 , input_length=14))

model.add(LSTM(200))

model.add(Dense(439,activation='softmax'))

model.build(input_shape=(None,14))
```

```python
model.compile(optimizer='adam',metrics=['accuracy'],loss='categorical_crossentropy')
model.summary()
```

```python
model.fit(x,y,epochs=100)
```
