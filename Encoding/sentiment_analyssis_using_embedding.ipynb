```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras import Sequential
from tensorflow.keras.layers import SimpleRNN , Dense , Embedding , Flatten
```

```python
(x_train,y_train) , (x_test,y_test) = imdb.load_data()
```

```python
x_train=pad_sequences(x_train,padding='post',maxlen=1000)
x_test=pad_sequences(x_test,padding='post',maxlen=1000)
```

```python
model1=Sequential()
model1.add(Embedding(
    input_dim=10000,
    output_dim=2,
    input_length=1000
))
model1.add(SimpleRNN(32,return_sequences=False))
model1.add(Dense(1,activation='sigmoid'))
model1.build(input_shape=(None, 1000))
model1.summary()
```

```python
model1.compile(optimizer='adam' , loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history=model1.fit(x_train,y_train , epochs=4 , validation_data=(x_test,y_test))
```
